```python
# 리스트와 튜플
# 리스트는 변경 가능한 자료형입니다.

my_list =[1,2,3]
my_list[0] = 10 # 리스트의 첫 번째 요소를 반환합니다.
my_list[1] = 20

len(my_list) # 리스트의 길이를 확인합니다.

my_list.append(4) # 리스트에 요소를 추가합니다. 1 2 3 4
my_list.remove(2) # 리스트에 특정 요소를 제거합니다.

my_list.sort() # 리스트를 오름차순으로 정렬합니다.
my_list[1:2] # 리스트의 특정 범위를 추출합니다.
print(my_list[1:2]) # 리스트의 현재 값을 출력합니다.



# 튜플은 변경 불가능한 자료형입니다.

my_tuple = (1,2,3)
# my_tuple[0] = 10 이 줄은 오류를 발생시킵니다.
len(my_tuple) # 튜플의 길이를 확인합니다.
my_tuple.count(2) # 튜플에서 특정 요소의 개수를 셉니다
my_tuple.index(3) # 튜플에서 특정 요소의 인덱스를 찾습니다.
```
