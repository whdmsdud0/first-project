```python
class PointWallet:
    def __init__(self, owner, points):
        self.owner = owner
        self.points = points

    def add_points(self, amount):
        self.points += amount
        print(f"{amount}포인트 적립")

    def use_points(self, amount):
        if self.points >= amount:
            self.points -= amount
            print(f"{amount}포인트 사용")
        else:
            print("포인트가 부족합니다.")

wallet = PointWallet("김민수", 5000)
#init을 실행합니다 -> wallet이라고 하는 인스턴스 안에 owner = 김민수, points = 5000
wallet.add_points(1000)
#add_points을 실행합니다 -> 인스턴스 안에 있는 points 값에다가 1000을 더해주고, 인스턴스 안에 있는 points 값을 갱신 => self points 값이 6천이 된다.

wallet.use_points(2000)
# use_points를 실행합니다. -> 인스턴스 안에 있는 points 값이 2천보다 큰지 확인하고, 크다면 인스턴스 안에 있는 point 값에다 2천 빼주고 값을 갱신 ㅍ

print(f"현재 포인트: {wallet.points}")
```

```python
# 문제 풀이
# 리스트에서 가장 길게 출석한 날 찾기 
# 조건 "o"이면 가장 긴 일수에 +1 
# 찾으면 반환


data: ["O", "O", "X", "O", "O", "O", "X"]

def solution(attendence):
    current = 0
    longest = 0

    for record in attendence: #0번째: "0"
        if record == "O":
            current += 1 # current += 1

            if current > longest: # longest = 1
                longest = current

        else:
            current = 0

    return longest

data1 = ["O", "O", "X", "O", "O", "O", "X"]
print(solution(data1))
```

```python
# 사용자의 참여 횟수를 저장할 빈 딕셔너리 만들기 > {}
# 반복하면서 사용자 이름을 하나씩 확인한다 > for
# 사용자 이름이 이미 딕셔너리에 있다면 > if ~ in
# 기존 참여 횟수에 1을 더한다 > += 1
# 처음 등장한 사용자라면 > else
# 완성된 딕셔너리를 반환한다 > return counts


def solution(participants):
    counts = {}
    for name in participants:
        if name in counts:
            counts[name] += 1

    else:
        counts[name] = 1

    return counts


print(solution(["minsu", "jisu", "minsu", "yuna", "minsu"]))
print(solution(["a", "b", "a", "b"]))
print(solution(["tom"]))
```

```python
# 요청을 들어온 순서대로 처리 deque 써라

from collections import deque # collections라는 곳에서 deque를 가져와서 사용할게.

def solution (commands):
    queue = deque()  
    # 이 식을 쓰는 이유는? 안쓰게 되면 파이썬이 deque를 인식 하지 못함
    # deque() 는 실제로 대기열을 만드는 기능. queue는 그 대기열을 저장해 놓을 이름. 
    # 즉 queue라는 이름의 대기열을 하나 만들겠다! queue 는 내가 마음대로 이름 정할 수 있음.
    
    for command in commands:
        if command == "RUN":
            queue.popleft() # 대기열 맨 앞의 작업을 꺼내서 처리한다

        else: # 파이썬에서 string.split()
              # ex) string_data = "안녕하세요. 저는 칸트가 좋습니다..?"
              # stirng_data.split() -> 다 띄어쓰기 처리해줌
            parts = command.split()
            task = parts[1]
            queue.append(task)

return list(queue)


  
```

```python
# heap, 여러 데이터 중에서 가장 작은 값이나 가장 큰 값을 반복적으로 빠르게 선택할 때 사용하는 자료구조

import heapq

numbers = []

heapq.heappush(numbers, 5) # heap에는 5
heapq.heappush(numbers, 2) # heap에는 5,2
heapq.heappush(numbers, 8) # heap에는 5,2,8

print(heapq.heappop(numbers)) # 2가 산출됨 
print(heapq.heappop(numbers)) # 5가 산출됨 
print(heapq.heappop(numbers)) # 마지막으로 8이 산출됨 
```

```python
# 가장 큰 값을 꺼낼때는 음수로 바꾸어 저장하는 방법
#2,5,8 -> -2, -5, -8

import heapq

numbers = []

heapq.heappush(numbers, -5)
heapq.heappush(numbers, -2)
heapq.heappush(numbers, -8)


print(-heapq.heappop(numbers)) # 가장 작은 8이 먼저 산출됨
print(-heapq.heappop(numbers)) # 그다음으로 5가 산출
print(-heapq.heappop(numbers)) # 마지막 2가 산출
```

```python
# heapq 사용 numbers는 4,1,3,2 

import heapq

def solution(priorities):   #def 함수이름 (함수 부분 실행시 사용할 변수):
    heap = []

    for priority in priorities:        #priorities에 있는 값을 하나씩 heap에 넣어라 
        heapq.heappush(heap, priority)

    result = []

    while heap:
        priority = heapq.heappop(heap)
        result.append(priority)

    return result


print(solution([4, 1, 3, 2]))    # print(f"{한수 내용} 출력")
print(solution([5, 2, 2, 8, 1]))
print(solution([7]))
```

```python
import heapq

def solution(priorities):
    heap = []

    for priority in priorities:
        heapq.heappush(heap, priority)

    result = []

    while heap:  #heap이 비어있지 않는 동안 계속 반복
        priority = heapq.heappop(heap) #heap은 가장 작은 값을 뽑아준다 = 가장 작은 값은 우선순위가 가장 높다
        result.append(priority)

    return result


print(solution([4,1,3,2]))
```
