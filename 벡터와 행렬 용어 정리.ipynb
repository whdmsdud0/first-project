### 벡터
### 벡터는 숫자들을 정해진 순서로 모아놓은 것
### ex)[5,2]
### 세영이의 출석은 5일, 과제는 2개를 제출했다면 이 두 가지 정보를 함께 묶고 싶을 때 [5,2]로 묶을 수 있다

### 벡터의 차원
###  [5,2]는 숫자가 2개이므로 2차원 벡터라고 부른다.
### [5,2,4]는 숫자가 3개이므로 3차원 벡터라고 부른다.

### 가중치 w
### 가중치는 각 입력값에 곱하는 값
### 어떤 입력이 점수 계산에 어떻게 반영되었는 지 조절함
### 참여 점수를 출석 하루당 2점, 과제 하나당 3점이라고 했을 때
### 세영이가 5일 출석했다면 5 x 2 = 10점을 받을 수 있다. 과제를 2개 냈다면 2 x 3 = 6점이다
### 그래서 16점을 받을 수 있었다.

### 편향 b
### 편향은 입력값에 가중치를 곱해 더한 결과에 추가로 더한 값
### 참여 점수를 계산할 때 모든 학생들에게 기본 점수 1점을 준다는 가정이라면
### 출석을 아예 하지 않은 하영은 0 x 2 + 0 x 3 + 1로 1점을 받을 수 있다.

### 파라미터
### 파라미터는 모델의 계산에 사용되며 학습으로 조정할 수 있는 값(가중치와 편향)
### 예를 들어 공부 시간으로 시험 점수를 예상한다고 가정해보면
### 예상 점수 = 공부시간 x 5 + 10 <- 이라는 기본 규칙에서
### 실제 2시간 공부한 학생의 예상 점수는 20점인데, 실제로 30점이라면 두 점수 차이를 계산하여
### 이를 조정하는 값

### 가중합과 출력 점수
### 가중합은 각 입력값에 대응하는 가중치를 곱한 뒤 모두 더한 값
### 아까 앞서 5 x 2 = 10점, 2 x 3 = 6점, 편향 1점 이면 17점
### 여기서 16점이 가중합이라고 부를 수 있다.

### 가중치 행렬
### 가중치 행렬은 여러 출력에 필요한 가중치들을 표로 모은 것
### w =[[2,1],[3,4]]라고 가정하면 같은 출석, 과제 기록으로 점수를 만든다고 할 때
### 첫 번째 점수는 출석당 2점, 과제당 3점, 둘째 점수는 출석당 1점, 과제당 4점으로 나타낼 수 있다.

### 노름 L1 Norm 
### L1 노름은 각 원소의 절대값을 구한 뒤 전부 더한 값
### 윤하가 오른쪽으로 3m 이동한 뒤 아래쪽으로는 4m 이동했다고 가정하면 이동 기록은 [3, -4]일 수 있다.
### 이걸 이동 거리로 계산해보면 |3|, |-4|, L1 노름을 이용하면 3 + 4 = 7m 이동했다고 볼 수 있다.

### 노름 L2 Norm
### L2 노름은 각 원소를 제곱하고 모두 더한 뒤, 제곱근을 구한 값
### 8강의실 출입문을 [0,0]으로 정하고 윤하가 오른쪽 6m , 위쪽 8m 지점에 있다고 가정을 해보면
### 출입문에서 윤하까지의 거리는 6 x 6 = 36, 8 x 8 = 64, 36 + 64 = 100 <- 루트를 써서 10m 정도 떨어져 있더는 것을 구할 수 있다.

### 차이 벡터
### 차이 벡터는 두 벡터에서 같은 위치의 값끼리 뺀 결과
### A = [3,4], B = [6,8]일 때, B - A = [6-3, 8-4] = [3,4]로 계산할 수 있다.
